```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation
import matplotlib as mpl
```

```python
ini = np.genfromtxt("Initial galaxy.txt")
a1 = np.genfromtxt("alpha=0.25.txt")
a2 = np.genfromtxt("alpha=0.5.txt")
a3 = np.genfromtxt("alpha=0.75.txt")
fi = np.genfromtxt("Final galaxy.txt")
```

```python
re = 2; Ie = 1; rmax = 3*re
mpl.rc('axes', labelsize=16, titlesize=16)
mpl.rcParams['font.family'] = 'Tahoma'
fig, ax = plt.subplots(1,5, figsize=(15,3), sharey=True)
norm = plt.cm.colors.LogNorm()
extent = (-rmax/re,rmax/re,-rmax/re,rmax/re)
ax[0].imshow(ini, cmap='hot', vmin=0, vmax=500, extent=extent)
ax[0].set_title(r"$\alpha = 0$")
ax[0].set_ylabel(r"Y [$R_e$]")
ax[1].imshow(a3 , cmap='hot', vmin=0, vmax=500, extent=extent)
ax[1].set_title(r"$\alpha = 0.25$")

ax[2].imshow(a2 , cmap='hot', vmin=0, vmax=500, extent=extent)
ax[2].set_title(r"$\alpha = 0.5$")
ax[3].imshow(a1 , cmap='hot', vmin=0, vmax=500, extent=extent)
ax[3].set_title(r"$\alpha = 0.75$")
test = ax[4].imshow(fi, cmap='hot', vmin=0, vmax=500, extent=extent)
ax[4].set_title(r"$\alpha = 1$")
cbar_ax = fig.add_axes([0.92,0.15,0.02,0.7])
fig.colorbar(test, cax=cbar_ax, label="Relative surface brightness")
for a in ax:
    a.set_facecolor('black')
    a.invert_yaxis()
plt.setp(ax.flat, xlabel=r"X [$R_e$]")
plt.savefig("Galaxy evolution", dpi=600)
```

```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import matplotlib as mpl

mpl.rcParams['font.family'] = 'Tahoma'

fig = plt.figure(figsize=(15, 8))  # Increased height for better balance
gs = gridspec.GridSpec(2, 3, height_ratios=[1, 1], width_ratios=[1, 1, 0.05])  # Adjusted layout to fix spacing

# First row (3 images)
ax0 = fig.add_subplot(gs[0, 0])
ax1 = fig.add_subplot(gs[0, 1])
ax2 = fig.add_subplot(gs[0, 2])

# Second row (2 images centered beneath middle plot)
ax3 = fig.add_subplot(gs[1, 0:2])  # Spanning columns 0 and 1
ax4 = fig.add_subplot(gs[1, 1:3])  # Spanning columns 1 and 2

axes = [ax0, ax1, ax2, ax3, ax4]
titles = [r"$\alpha = 0$", r"$\alpha = 0.25$", r"$\alpha = 0.5$",
          r"$\alpha = 0.75$", r"$\alpha = 1$"]
images = [ini, a3, a2, a1, fi]

norm = plt.cm.colors.LogNorm()

for ax, img, title in zip(axes, images, titles):
    im = ax.imshow(img, cmap='hot', vmin=0, vmax=500)
    ax.set_title(title, fontsize=14)
    ax.set_facecolor('black')
    ax.invert_yaxis()

# Adjust layout
for ax in [ax0, ax1, ax2]:
    ax.set_ylabel(r"Y [$R_e$]")

plt.setp(axes, xlabel=r"X [$R_e$]")

# Colorbar
cbar_ax = fig.add_axes([0.92, 0.15, 0.02, 0.7])  # Colorbar on the right
fig.colorbar(im, cax=cbar_ax, label="Relative surface brightness")
plt.savefig("Galaxy_evolution.png", dpi=600)
plt.show()
```
